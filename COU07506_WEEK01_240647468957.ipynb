```python


import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile
plt.rcParams["figure.figsize"] = (9, 3.5)

# Define discrete-time index
n = np.arange(-5, 11)

# Generate signals
delta = (n == 0).astype(float)  # Unit impulse
step = (n >= 0).astype(float)   # Unit step
ramp = n * step                 # Unit ramp
expo = 0.8**n * step            # Exponential signal

# Plot the four signals
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ[n]", "u[n]", "r[n] = n u[n]", "(0.8)^n u[n]"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

Task 1.2 (a)

```python


import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile
plt.rcParams["figure.figsize"] = (9, 3.5)


n = np.arange(-5, 11)

u = (n >= 0).astype(float)

# Three exponential signals
a1 = 0.5
a2 = -0.8
a3 = 1.1

x1 = (a1 ** n) * u
x2 = (a2 ** n) * u
x3 = (a3 ** n) * u

# Plot using subplots
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].stem(n, x1)
axes[0].set_title("a = 0.5 (Decaying)")
axes[0].set_xlabel("n")
axes[0].set_ylabel("Amplitude")
axes[0].grid(True)

axes[1].stem(n, x2)
axes[1].set_title("a = -0.8 (Alternating)")
axes[1].set_xlabel("n")
axes[1].set_ylabel("Amplitude")
axes[1].grid(True)

axes[2].stem(n, x3)
axes[2].set_title("a = 1.1 (Growing)")
axes[2].set_xlabel("n")
axes[2].set_ylabel("Amplitude")
axes[2].grid(True)

plt.tight_layout()
plt.show()
```

# 1.The sampling rate is the number of audio samples recorded per second. It is measured in Hertz (Hz)
To calculate the number of samples needed to store one minute of mono audio:
\[
N=f_s\times T
\]

Where:
- \(N\) = total number of samples.
- \(f_s\) = sampling rate.
- \(T\) = duration in seconds.

# 2.A stem plot is more appropriate because \(x(n)\) represents a discrete-time signal, which consists of individual samples at specific integer values of \(n\).

A stem plot displays each sample separately using a vertical line and a marker. A continuous line may incorrectly suggest that the signal exists at every point between samples.

Therefore, a stem plot clearly represents the discrete nature of digital signals.

# 3.For \(a=1.1\), the exponential grows. Which property of a system would such a signal violate if it were an impulse response?
It would violate the BIBO stability property (Bounded-Input Bounded-Output stability).

A system is BIBO stable if every bounded input produces a bounded output.
When \(a=1.1\), the exponential impulse response grows with time instead of decaying. Its absolute sum does not converge.

Therefore, the system would be unstable (not BIBO stable).
